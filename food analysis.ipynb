
# Task 3 — Data Quality Summary and Cleaning Rules

## 1. Data Quality Issues Found

1. **Missing values:** Several columns contain missing values. Discounts and Offers has 242 missing values (23.96%), while other columns such as Order Value and Delivery Fee also have missing records. This can affect financial calculations and business reports.

2. **Duplicate records:** The dataset contains 10 fully duplicated rows. These can cause the same order to be counted more than once, leading to incorrect business totals.

3. **Repeated Order IDs:** There are 10 unique Order IDs that each appear twice. This may indicate duplicate orders or repeated records. These records must be checked before removing anything.

4. **Incorrect date data types:** Order Date and Time and Delivery Date and Time are initially stored as strings instead of datetime values. This makes date-based analysis and delivery-time calculations difficult.

5. **Missing financial and categorical information:** Missing fees, refunds, payment methods, and offer details can reduce the accuracy of financial and customer-behaviour analysis.

## 2. Data Cleaning Rules

| Data-quality issue | Your chosen handling rule | Reason / business assumption |
|---|---|---|
| Exact duplicate rows | Remove one copy of each fully identical row using `drop_duplicates()`. | Prevents the same order record from being counted twice. |
| Repeated Order IDs | Compare records with the same Order ID. Remove a record only if it is an exact duplicate. Keep different records for investigation. | An Order ID should identify an order. Different records with the same ID may indicate a data-entry or system issue. |
| Missing Order Value | Keep missing values as NaN and exclude those records from Order Value calculations. | We cannot assume an unknown order value is zero or estimate the actual amount without evidence. |
| Missing Delivery Fee | Keep missing values as NaN and exclude them from Delivery Fee calculations. | A missing fee does not mean the customer paid zero. |
| Missing Commission Fee | Keep missing values as NaN and exclude them from Commission Fee calculations. | Filling with zero could understate the company's commission income. |
| Missing Payment Processing Fee | Keep missing values as NaN and exclude them from processing-fee calculations. | A missing processing fee does not prove that no fee was charged. |
| Missing Refunds/Chargebacks | Keep missing values as NaN. Do not replace them with zero. | A missing refund amount does not necessarily mean that no refund occurred. |
| Missing Payment Method | Replace missing categorical values with `Unknown`. | The payment method is not recorded, so we should not guess how the customer paid. |
| Missing Discounts and Offers | Replace missing values with `Not Specified`. | This means the offer information is unavailable; it does not mean no discount was applied. |
| Invalid timestamps | Convert date columns to datetime and identify invalid values. Treat invalid timestamps as missing for time-based analysis. | Invalid dates can produce incorrect delivery-time and date-based reports. |
| Impossible delivery durations | Identify negative delivery durations and exclude them from delivery-time calculations until investigated. | A delivery cannot normally be completed before the order is placed. |

**Important assumption:** The original `raw_df` will remain unchanged. All cleaning operations will be performed on `work_df`.

### Validation 6 – Verify Delivery-Time Calculation

This check manually calculates delivery duration for a few records using the order and delivery timestamps. The manually calculated values are compared with the derived delivery-time column.

```python
# Select five records for manual verification
sample_records = work_df[
    [
        "Order ID",
        "Order Date and Time",
        "Delivery Date and Time",
        "Delivery_Time_Minutes"
    ]
].head(5).copy()

# Manually calculate delivery time
sample_records["Manual_Delivery_Time"] = (
    sample_records["Delivery Date and Time"]
    - sample_records["Order Date and Time"]
).dt.total_seconds() / 60

# Compare calculated and manual values
sample_records["Calculation_Correct"] = np.isclose(
    sample_records["Delivery_Time_Minutes"],
    sample_records["Manual_Delivery_Time"]
)

# Display results
display(sample_records)

print(
    "\nAll calculations correct:",
    sample_records["Calculation_Correct"].all()
)
```

### Validation 7 – Verify Platform Net Revenue Calculation

This check manually recalculates platform net revenue for five records with all required financial values. The manually calculated results are compared with the derived revenue column.

```python
# Select records with all required financial values
revenue_sample = work_df.dropna(
    subset=[
        "Commission Fee",
        "Delivery Fee",
        "Payment Processing Fee",
        "Refunds/Chargebacks",
        "Platform_Net_Revenue"
    ]
).head(5).copy()

# Manually calculate platform net revenue
revenue_sample["Manual_Net_Revenue"] = (
    revenue_sample["Commission Fee"]
    + revenue_sample["Delivery Fee"]
    - revenue_sample["Payment Processing Fee"]
    - revenue_sample["Refunds/Chargebacks"]
)

# Compare calculated and manual revenue
revenue_sample["Calculation_Correct"] = np.isclose(
    revenue_sample["Platform_Net_Revenue"],
    revenue_sample["Manual_Net_Revenue"]
)

# Display results
display(
    revenue_sample[
        [
            "Order ID",
            "Commission Fee",
            "Delivery Fee",
            "Payment Processing Fee",
            "Refunds/Chargebacks",
            "Platform_Net_Revenue",
            "Manual_Net_Revenue",
            "Calculation_Correct"
        ]
    ]
)

print(
    "\nAll revenue calculations correct:",
    revenue_sample["Calculation_Correct"].all()
)
```
